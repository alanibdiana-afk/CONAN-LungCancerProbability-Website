```python
import numpy as np
import pandas as pd

nodule = df[df["class_name"] == "Nodule/Mass"].copy()

def calculate_iou(box1, box2):
    x1_min, y1_min, x1_max, y1_max = box1
    x2_min, y2_min, x2_max, y2_max = box2

    inter_xmin = max(x1_min, x2_min)
    inter_ymin = max(y1_min, y2_min)
    inter_xmax = min(x1_max, x2_max)
    inter_ymax = min(y1_max, y2_max)

    inter_w = max(0, inter_xmax - inter_xmin)
    inter_h = max(0, inter_ymax - inter_ymin)
    intersection = inter_w * inter_h

    area1 = max(0, x1_max - x1_min) * max(0, y1_max - y1_min)
    area2 = max(0, x2_max - x2_min) * max(0, y2_max - y2_min)

    union = area1 + area2 - intersection

    if union == 0:
        return 0.0

    return intersection / union


# Calculate pairwise IoU within each image
ious = []

for image_id, group in nodule.groupby("image_id"):

    records = group[
        ["rad_id", "x_min", "y_min", "x_max", "y_max"]
    ].to_dict("records")

    for i in range(len(records)):
        for j in range(i + 1, len(records)):

            box1 = (
                records[i]["x_min"],
                records[i]["y_min"],
                records[i]["x_max"],
                records[i]["y_max"]
            )

            box2 = (
                records[j]["x_min"],
                records[j]["y_min"],
                records[j]["x_max"],
                records[j]["y_max"]
            )

            iou = calculate_iou(box1, box2)

            ious.append({
                "image_id": image_id,
                "rad1": records[i]["rad_id"],
                "rad2": records[j]["rad_id"],
                "iou": iou
            })

iou_df = pd.DataFrame(ious)

print("Number of Nodule/Mass annotation pairs:", len(iou_df))

print("\nIoU statistics:")
display(iou_df["iou"].describe())
```

```python
print("\nIoU categories:")

iou_categories = pd.cut(
    iou_df["iou"],
    bins=[-0.001, 0, 0.25, 0.50, 0.75, 1.0],
    labels=[
        "No overlap",
        "Low overlap",
        "Moderate overlap",
        "High overlap",
        "Very high overlap"
    ]
)

display(
    iou_categories
    .value_counts()
    .sort_index()
    .rename_axis("IoU category")
    .reset_index(name="annotation_pairs")
)
```

```python
# --------------------------------------------------
# RESTART-SAFE: Load annotations + image dimensions
# --------------------------------------------------

import os
import numpy as np
import pandas as pd
import pydicom

DATA_DIR = (
    "/kaggle/input/competitions/"
    "vinbigdata-chest-xray-abnormalities-detection"
)

TRAIN_DIR = os.path.join(DATA_DIR, "train")

# Always reload the original annotation CSV
df = pd.read_csv(
    os.path.join(DATA_DIR, "train.csv")
)

print("Original annotation table:", df.shape)

# --------------------------------------------------
# Read image dimensions
# --------------------------------------------------

image_dimensions = {}

dicom_files = [
    f for f in os.listdir(TRAIN_DIR)
    if f.endswith(".dicom")
]

print("DICOM files found:", len(dicom_files))

for filename in dicom_files:

    image_id = filename.replace(".dicom", "")
    path = os.path.join(TRAIN_DIR, filename)

    try:
        ds = pydicom.dcmread(
            path,
            stop_before_pixels=True
        )

        image_dimensions[image_id] = {
            "height": int(ds.Rows),
            "width": int(ds.Columns),
            "photometric": getattr(
                ds,
                "PhotometricInterpretation",
                None
            )
        }

    except Exception as e:
        print("Failed:", filename, e)

print(
    "Images with dimensions:",
    len(image_dimensions)
)

# --------------------------------------------------
# Convert dimensions to DataFrame
# --------------------------------------------------

dims = (
    pd.DataFrame.from_dict(
        image_dimensions,
        orient="index"
    )
    .rename_axis("image_id")
    .reset_index()
)

print("\nDimension table:")
display(dims.head())

# --------------------------------------------------
# Merge
# --------------------------------------------------

df = df.merge(
    dims,
    on="image_id",
    how="left",
    validate="many_to_one"
)

print("\nMerged dataset:", df.shape)

print("\nMissing image dimensions:")
print(
    df[
        ["height", "width", "photometric"]
    ].isnull().sum()
)

display(df.head())
```

```python
# --------------------------------------------------
# CREATE NORMALIZED BOUNDING-BOX FEATURES
# --------------------------------------------------

bbox_mask = (
    df["x_min"].notna()
    & df["y_min"].notna()
    & df["x_max"].notna()
    & df["y_max"].notna()
)

df["box_width"] = np.where(
    bbox_mask,
    df["x_max"] - df["x_min"],
    np.nan
)

df["box_height"] = np.where(
    bbox_mask,
    df["y_max"] - df["y_min"],
    np.nan
)

df["box_area"] = (
    df["box_width"] *
    df["box_height"]
)

# Normalized dimensions
df["box_width_norm"] = (
    df["box_width"] / df["width"]
)

df["box_height_norm"] = (
    df["box_height"] / df["height"]
)

df["box_area_norm"] = (
    df["box_area"] /
    (df["width"] * df["height"])
)

# Normalized center coordinates
df["center_x_norm"] = (
    ((df["x_min"] + df["x_max"]) / 2)
    / df["width"]
)

df["center_y_norm"] = (
    ((df["y_min"] + df["y_max"]) / 2)
    / df["height"]
)

print("Bounding-box features created.")

display(
    df[
        [
            "image_id",
            "class_name",
            "box_width_norm",
            "box_height_norm",
            "box_area_norm",
            "center_x_norm",
            "center_y_norm"
        ]
    ].head(20)
)
```

```python
# ============================================================
# CONAN IMAGING — STEP 2
# Load VinBigData annotations + DICOM image dimensions
# ============================================================

import os
import numpy as np
import pandas as pd
import pydicom

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

DATA_DIR = (
    "/kaggle/input/competitions/"
    "vinbigdata-chest-xray-abnormalities-detection"
)

TRAIN_DIR = os.path.join(DATA_DIR, "train")
CSV_PATH = os.path.join(DATA_DIR, "train.csv")

print("DATA DIRECTORY:")
print(DATA_DIR)

print("\nTRAIN DIRECTORY:")
print(TRAIN_DIR)

# ------------------------------------------------------------
# 2. Load original VinBigData annotations
# ------------------------------------------------------------

df = pd.read_csv(CSV_PATH)

print("\nOriginal annotation dataset:")
print("Rows:", len(df))
print("Columns:", len(df.columns))
print(df.columns.tolist())

# ------------------------------------------------------------
# 3. Find all DICOM files
# ------------------------------------------------------------

dicom_files = [
    f for f in os.listdir(TRAIN_DIR)
    if f.lower().endswith(".dicom")
]

print("\nDICOM files found:", len(dicom_files))

# ------------------------------------------------------------
# 4. Read image dimensions
# ------------------------------------------------------------

image_dimensions = {}

failed_files = []

for i, filename in enumerate(dicom_files):

    image_id = filename.rsplit(".", 1)[0]

    path = os.path.join(TRAIN_DIR, filename)

    try:

        ds = pydicom.dcmread(
            path,
            stop_before_pixels=True
        )

        image_dimensions[image_id] = {
            "height": int(ds.Rows),
            "width": int(ds.Columns),
            "photometric": getattr(
                ds,
                "PhotometricInterpretation",
                None
            )
        }

    except Exception as e:

        failed_files.append(
            (filename, str(e))
        )

# ------------------------------------------------------------
# 5. Create image-dimension table
# ------------------------------------------------------------

dims = pd.DataFrame(
    [
        {
            "image_id": image_id,
            "height": values["height"],
            "width": values["width"],
            "photometric": values["photometric"]
        }
        for image_id, values
        in image_dimensions.items()
    ]
)

print("\nImage dimension table:")
print("Rows:", len(dims))

display(dims.head())

# ------------------------------------------------------------
# 6. Check dimension reading
# ------------------------------------------------------------

print("\nFailed DICOM files:", len(failed_files))

if failed_files:
    display(
        pd.DataFrame(
            failed_files,
            columns=["filename", "error"]
        ).head()
    )

# ------------------------------------------------------------
# 7. Check duplicate image IDs
# ------------------------------------------------------------

print(
    "\nDuplicate image IDs in dimension table:",
    dims["image_id"].duplicated().sum()
)

# ------------------------------------------------------------
# 8. Merge annotations with dimensions
# ------------------------------------------------------------

df = df.merge(
    dims,
    on="image_id",
    how="left",
    validate="many_to_one"
)

print("\nMerged dataset:")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumns:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 9. Verify missing dimensions
# ------------------------------------------------------------

print("\nMissing dimensions:")

display(
    df[
        ["height", "width", "photometric"]
    ].isnull().sum()
)

# ------------------------------------------------------------
# 10. Verify image coverage
# ------------------------------------------------------------

annotation_images = set(
    df["image_id"].unique()
)

dimension_images = set(
    dims["image_id"].unique()
)

print("\nUnique annotation images:",
      len(annotation_images))

print("Unique dimension images:",
      len(dimension_images))

print(
    "Annotation images without dimensions:",
    len(annotation_images - dimension_images)
)

# ------------------------------------------------------------
# 11. Final preview
# ------------------------------------------------------------

print("\nFinal preview:")

display(
    df[
        [
            "image_id",
            "class_name",
            "rad_id",
            "x_min",
            "y_min",
            "x_max",
            "y_max",
            "height",
            "width",
            "photometric"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CONAN IMAGING — STEP 3
# Create normalized bounding-box features
# ============================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Identify rows that actually contain bounding boxes
# ------------------------------------------------------------

bbox_mask = (
    df["x_min"].notna()
    & df["y_min"].notna()
    & df["x_max"].notna()
    & df["y_max"].notna()
)

print("Total annotation rows:", len(df))
print("Rows with bounding boxes:", bbox_mask.sum())
print("Rows without bounding boxes:", (~bbox_mask).sum())

# ------------------------------------------------------------
# 2. Raw bounding-box dimensions
# ------------------------------------------------------------

df["box_width"] = np.where(
    bbox_mask,
    df["x_max"] - df["x_min"],
    np.nan
)

df["box_height"] = np.where(
    bbox_mask,
    df["y_max"] - df["y_min"],
    np.nan
)

df["box_area"] = (
    df["box_width"] *
    df["box_height"]
)

# ------------------------------------------------------------
# 3. Normalized bounding-box dimensions
# ------------------------------------------------------------

df["box_width_norm"] = (
    df["box_width"] / df["width"]
)

df["box_height_norm"] = (
    df["box_height"] / df["height"]
)

df["box_area_norm"] = (
    df["box_area"] /
    (df["width"] * df["height"])
)

# ------------------------------------------------------------
# 4. Bounding-box center
# ------------------------------------------------------------

df["center_x"] = (
    df["x_min"] + df["x_max"]
) / 2

df["center_y"] = (
    df["y_min"] + df["y_max"]
) / 2

# ------------------------------------------------------------
# 5. Normalize center to 0–1
# ------------------------------------------------------------

df["center_x_norm"] = (
    df["center_x"] / df["width"]
)

df["center_y_norm"] = (
    df["center_y"] / df["height"]
)

# ------------------------------------------------------------
# 6. Basic validity checks
# ------------------------------------------------------------

print("\nBounding-box feature ranges:")

display(
    df.loc[bbox_mask, [
        "box_width_norm",
        "box_height_norm",
        "box_area_norm",
        "center_x_norm",
        "center_y_norm"
    ]].describe()
)

# ------------------------------------------------------------
# 7. Check for invalid normalized values
# ------------------------------------------------------------

feature_cols = [
    "box_width_norm",
    "box_height_norm",
    "box_area_norm",
    "center_x_norm",
    "center_y_norm"
]

print("\nInvalid values:")

for col in feature_cols:
    invalid = (
        (df[col] < 0) |
        (df[col] > 1)
    ).sum()

    print(f"{col}: {invalid}")

# ------------------------------------------------------------
# 8. Preview
# ------------------------------------------------------------

print("\nFeature preview:")

display(
    df.loc[
        bbox_mask,
        [
            "image_id",
            "class_name",
            "rad_id",
            "box_width_norm",
            "box_height_norm",
            "box_area_norm",
            "center_x_norm",
            "center_y_norm"
        ]
    ].head(20)
)
```

```python
# ============================================================
# CONAN IMAGING — STEP 4
# Build image-level Nodule/Mass features
# ============================================================

# ------------------------------------------------------------
# 1. Select Nodule/Mass annotations
# ------------------------------------------------------------

nm = df[
    df["class_name"] == "Nodule/Mass"
].copy()

print("Nodule/Mass annotation rows:", len(nm))
print("Unique Nodule/Mass images:", nm["image_id"].nunique())

# ------------------------------------------------------------
# 2. Number of unique radiologists per image
# ------------------------------------------------------------

nm_radiologists = (
    nm.groupby("image_id")["rad_id"]
    .nunique()
    .rename("nodule_mass_radiologists")
)

# ------------------------------------------------------------
# 3. Annotation count per image
# ------------------------------------------------------------

nm_annotations = (
    nm.groupby("image_id")
    .size()
    .rename("nodule_mass_annotation_count")
)

# ------------------------------------------------------------
# 4. Radiologist agreement
# ------------------------------------------------------------

nm_agreement = (
    nm_radiologists / 3.0
).rename("nodule_mass_agreement")

# ------------------------------------------------------------
# 5. Aggregate normalized box measurements
# ------------------------------------------------------------

nm_size_location = (
    nm.groupby("image_id")
    .agg(
        nodule_mass_max_width=(
            "box_width_norm",
            "max"
        ),

        nodule_mass_mean_width=(
            "box_width_norm",
            "mean"
        ),

        nodule_mass_max_height=(
            "box_height_norm",
            "max"
        ),

        nodule_mass_mean_height=(
            "box_height_norm",
            "mean"
        ),

        nodule_mass_max_area=(
            "box_area_norm",
            "max"
        ),

        nodule_mass_mean_area=(
            "box_area_norm",
            "mean"
        ),

        nodule_mass_mean_x=(
            "center_x_norm",
            "mean"
        ),

        nodule_mass_mean_y=(
            "center_y_norm",
            "mean"
        ),

        nodule_mass_min_x=(
            "center_x_norm",
            "min"
        ),

        nodule_mass_max_x=(
            "center_x_norm",
            "max"
        ),

        nodule_mass_min_y=(
            "center_y_norm",
            "min"
        ),

        nodule_mass_max_y=(
            "center_y_norm",
            "max"
        )
    )
)

# ------------------------------------------------------------
# 6. Combine Nodule/Mass features
# ------------------------------------------------------------

nodule_features = pd.concat(
    [
        nm_radiologists,
        nm_annotations,
        nm_agreement,
        nm_size_location
    ],
    axis=1
).reset_index()

# ------------------------------------------------------------
# 7. Explicit presence flag
# ------------------------------------------------------------

nodule_features["nodule_mass_present"] = 1

# ------------------------------------------------------------
# 8. Display result
# ------------------------------------------------------------

print("\nNodule/Mass image-level dataset:")
print("Rows:", len(nodule_features))
print("Columns:", len(nodule_features.columns))

display(nodule_features.head(20))
```

```python
# ============================================================
# VERIFY NODULE/MASS AGREEMENT
# ============================================================

print("Radiologists identifying Nodule/Mass:")

display(
    nodule_features[
        "nodule_mass_radiologists"
    ]
    .value_counts()
    .sort_index()
    .rename_axis("radiologists")
    .reset_index(name="images")
)

print("\nAgreement distribution:")

display(
    nodule_features[
        "nodule_mass_agreement"
    ]
    .value_counts()
    .sort_index()
    .rename_axis("agreement")
    .reset_index(name="images")
)

print("\nAnnotation count distribution:")

display(
    nodule_features[
        "nodule_mass_annotation_count"
    ].describe()
)
```

```python
# ============================================================
# CONAN IMAGING — COMPLETE NODULE/MASS FEATURE TABLE
# ============================================================

# Get one row per image from the original dataset
image_base = (
    df[
        [
            "image_id",
            "height",
            "width",
            "photometric"
        ]
    ]
    .drop_duplicates("image_id")
    .copy()
)

print("Unique images:", len(image_base))

# Merge Nodule/Mass features
imaging_nm = image_base.merge(
    nodule_features,
    on="image_id",
    how="left"
)

# ------------------------------------------------------------
# Fill values for images without Nodule/Mass
# ------------------------------------------------------------

zero_columns = [
    "nodule_mass_radiologists",
    "nodule_mass_annotation_count",
    "nodule_mass_agreement",
    "nodule_mass_present"
]

for col in zero_columns:
    imaging_nm[col] = (
        imaging_nm[col]
        .fillna(0)
    )

# Size/location features are undefined when
# no Nodule/Mass exists.
# For a risk model, zero provides a clear
# "no lesion evidence" representation.

feature_zero_columns = [
    "nodule_mass_max_width",
    "nodule_mass_mean_width",
    "nodule_mass_max_height",
    "nodule_mass_mean_height",
    "nodule_mass_max_area",
    "nodule_mass_mean_area",
    "nodule_mass_mean_x",
    "nodule_mass_mean_y",
    "nodule_mass_min_x",
    "nodule_mass_max_x",
    "nodule_mass_min_y",
    "nodule_mass_max_y"
]

for col in feature_zero_columns:
    imaging_nm[col] = (
        imaging_nm[col]
        .fillna(0)
    )

# ------------------------------------------------------------
# Final verification
# ------------------------------------------------------------

print("\nFINAL NODULE/MASS TABLE")
print("-----------------------")
print("Rows:", len(imaging_nm))
print("Columns:", len(imaging_nm.columns))

print("\nNodule/Mass presence:")
display(
    imaging_nm[
        "nodule_mass_present"
    ]
    .value_counts()
    .sort_index()
    .rename_axis("present")
    .reset_index(name="images")
)

print("\nMissing values:")
display(
    imaging_nm.isnull().sum()
)

print("\nPreview:")
display(
    imaging_nm.head(10)
)
```

```python
# ============================================================
# CONAN IMAGING — STEP 5
# Build complete image-level radiographic feature matrix
# ============================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Define actual abnormality classes
# ------------------------------------------------------------

abnormalities = [
    "Aortic enlargement",
    "Cardiomegaly",
    "Pleural thickening",
    "ILD",
    "Nodule/Mass",
    "Pulmonary fibrosis",
    "Lung Opacity",
    "Atelectasis",
    "Other lesion",
    "Infiltration",
    "Pleural effusion",
    "Calcification",
    "Consolidation",
    "Pneumothorax"
]

print("Number of abnormality classes:", len(abnormalities))
print(abnormalities)

# ------------------------------------------------------------
# 2. Start with exactly one row per X-ray
# ------------------------------------------------------------

image_features = (
    df[
        [
            "image_id",
            "height",
            "width",
            "photometric"
        ]
    ]
    .drop_duplicates("image_id")
    .copy()
)

print("\nInitial image table:")
print("Rows:", len(image_features))

# ------------------------------------------------------------
# 3. Process each abnormality
# ------------------------------------------------------------

for abnormality in abnormalities:

    # Create a safe column prefix
    prefix = (
        abnormality
        .lower()
        .replace("/", "_")
        .replace(" ", "_")
    )

    # Select annotations for this abnormality
    subset = df[
        df["class_name"] == abnormality
    ].copy()

    if len(subset) == 0:
        print(f"WARNING: No annotations for {abnormality}")
        continue

    # --------------------------------------------------------
    # Radiologist count
    # --------------------------------------------------------

    radiologist_count = (
        subset.groupby("image_id")["rad_id"]
        .nunique()
        .rename(
            f"{prefix}_radiologists"
        )
    )

    # --------------------------------------------------------
    # Annotation count
    # --------------------------------------------------------

    annotation_count = (
        subset.groupby("image_id")
        .size()
        .rename(
            f"{prefix}_annotation_count"
        )
    )

    # --------------------------------------------------------
    # Agreement
    #
    # Each VinBigData image was evaluated by 3 radiologists.
    # --------------------------------------------------------

    agreement = (
        radiologist_count / 3.0
    ).rename(
        f"{prefix}_agreement"
    )

    # --------------------------------------------------------
    # Spatial and size features
    # --------------------------------------------------------

    measurements = (
        subset.groupby("image_id")
        .agg(

            max_width=(
                "box_width_norm",
                "max"
            ),

            mean_width=(
                "box_width_norm",
                "mean"
            ),

            max_height=(
                "box_height_norm",
                "max"
            ),

            mean_height=(
                "box_height_norm",
                "mean"
            ),

            max_area=(
                "box_area_norm",
                "max"
            ),

            mean_area=(
                "box_area_norm",
                "mean"
            ),

            mean_x=(
                "center_x_norm",
                "mean"
            ),

            mean_y=(
                "center_y_norm",
                "mean"
            )
        )
    )

    # Rename measurement columns
    measurements = measurements.rename(
        columns={
            "max_width":
                f"{prefix}_max_width",

            "mean_width":
                f"{prefix}_mean_width",

            "max_height":
                f"{prefix}_max_height",

            "mean_height":
                f"{prefix}_mean_height",

            "max_area":
                f"{prefix}_max_area",

            "mean_area":
                f"{prefix}_mean_area",

            "mean_x":
                f"{prefix}_mean_x",

            "mean_y":
                f"{prefix}_mean_y"
        }
    )

    # --------------------------------------------------------
    # Combine this abnormality's features
    # --------------------------------------------------------

    abnormality_features = pd.concat(
        [
            radiologist_count,
            annotation_count,
            agreement,
            measurements
        ],
        axis=1
    ).reset_index()

    # --------------------------------------------------------
    # Presence indicator
    # --------------------------------------------------------

    abnormality_features[
        f"{prefix}_present"
    ] = 1

    # --------------------------------------------------------
    # Merge into master image table
    # --------------------------------------------------------

    image_features = image_features.merge(
        abnormality_features,
        on="image_id",
        how="left",
        validate="one_to_one"
    )

    # --------------------------------------------------------
    # Fill absence with zero
    # --------------------------------------------------------

    new_columns = [
        c for c in abnormality_features.columns
        if c != "image_id"
    ]

    image_features[new_columns] = (
        image_features[new_columns]
        .fillna(0)
    )

    print(
        f"{abnormality:25s} → "
        f"{len(subset):5d} annotations | "
        f"{subset['image_id'].nunique():5d} images"
    )

# ------------------------------------------------------------
# 4. Final dimensions
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL RADIOGRAPHIC FEATURE MATRIX")
print("=" * 70)

print("Images:", len(image_features))
print("Features:", len(image_features.columns))

print("\nShape:")
print(image_features.shape)

# ------------------------------------------------------------
# 5. Verify no duplicate image IDs
# ------------------------------------------------------------

print(
    "\nDuplicate image IDs:",
    image_features["image_id"].duplicated().sum()
)

# ------------------------------------------------------------
# 6. Missing-value check
# ------------------------------------------------------------

missing = (
    image_features
    .isnull()
    .sum()
)

print("\nColumns with missing values:")
display(
    missing[missing > 0]
)

# ------------------------------------------------------------
# 7. Display presence counts
# ------------------------------------------------------------

presence_columns = [
    c for c in image_features.columns
    if c.endswith("_present")
]

presence_summary = []

for col in presence_columns:

    presence_summary.append({
        "feature": col,
        "images_present":
            int(image_features[col].sum()),
        "percentage":
            float(
                image_features[col].mean() * 100
            )
    })

presence_summary = pd.DataFrame(
    presence_summary
).sort_values(
    "images_present",
    ascending=False
)

print("\nAbnormality presence:")
display(presence_summary)

# ------------------------------------------------------------
# 8. Preview
# ------------------------------------------------------------

print("\nFeature matrix preview:")
display(
    image_features.head()
)
```

```python
# ================================================================
# CONAN — JSRT STRATIFIED 5-FOLD CROSS-VALIDATION SETUP
# ================================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import StratifiedKFold

print("=" * 70)
print("CONAN — JSRT 5-FOLD CROSS-VALIDATION")
print("=" * 70)

# ------------------------------------------------
# 1. Load cleaned JSRT dataset
# ------------------------------------------------

input_path = "/kaggle/working/conan_jsrt_cancer_images.csv"

df = pd.read_csv(input_path)

print("\nLoaded:")
print(input_path)

print("\nDataset shape:")
print(df.shape)


# ------------------------------------------------
# 2. Validate target
# ------------------------------------------------

assert "lung_cancer" in df.columns
assert "image_path" in df.columns

print("\nTarget distribution:")
print(df["lung_cancer"].value_counts().sort_index())


# ------------------------------------------------
# 3. Create stratified 5-fold split
# ------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

df["fold"] = -1

for fold, (train_idx, val_idx) in enumerate(
    skf.split(df, df["lung_cancer"])
):

    df.loc[val_idx, "fold"] = fold


# ------------------------------------------------
# 4. Validate folds
# ------------------------------------------------

print("\n" + "=" * 70)
print("FOLD DISTRIBUTION")
print("=" * 70)

for fold in range(5):

    fold_data = df[df["fold"] == fold]

    malignant = int(
        (fold_data["lung_cancer"] == 1).sum()
    )

    benign = int(
        (fold_data["lung_cancer"] == 0).sum()
    )

    print(
        f"Fold {fold}: "
        f"{len(fold_data):3d} images | "
        f"Benign = {benign:2d} | "
        f"Malignant = {malignant:2d}"
    )


# ------------------------------------------------
# 5. Check every image belongs to exactly one fold
# ------------------------------------------------

assert df["fold"].isin(range(5)).all()
assert df["study_id"].is_unique
assert df["image_path"].notna().all()

print("\n✓ Every image has exactly one fold")
print("✓ No duplicate study IDs")
print("✓ All image paths exist in the dataset")


# ------------------------------------------------
# 6. Check class balance
# ------------------------------------------------

print("\n" + "=" * 70)
print("CLASS DISTRIBUTION BY FOLD")
print("=" * 70)

fold_table = pd.crosstab(
    df["fold"],
    df["lung_cancer"]
)

fold_table.columns = [
    "Benign" if c == 0 else "Malignant"
    for c in fold_table.columns
]

print(fold_table)


# ------------------------------------------------
# 7. Save
# ------------------------------------------------

output_path = "/kaggle/working/conan_jsrt_5fold.csv"

df.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(output_path)
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())
```

```python
# ================================================================
# CONAN — JSRT IMAGE PREPROCESSING + RESNET-50 FEATURE EXTRACTION
# ================================================================

import os
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torchvision
from torchvision import models, transforms

print("=" * 70)
print("CONAN — JSRT CNN FEATURE EXTRACTION")
print("=" * 70)

# ------------------------------------------------
# 1. Load the 5-fold dataset
# ------------------------------------------------

input_path = "/kaggle/working/conan_jsrt_5fold.csv"

df = pd.read_csv(input_path)

print("\nLoaded:")
print(input_path)

print("\nDataset:")
print(df.shape)


# ------------------------------------------------
# 2. Select device
# ------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nDevice:")
print(device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


# ------------------------------------------------
# 3. Load pretrained ResNet-50
# ------------------------------------------------

print("\nLoading pretrained ResNet-50...")

weights = models.ResNet50_Weights.DEFAULT

resnet = models.resnet50(
    weights=weights
)

# Remove final classification layer.
# Output becomes the 2048-dimensional image embedding.

feature_extractor = torch.nn.Sequential(
    *list(resnet.children())[:-1]
)

feature_extractor = feature_extractor.to(device)
feature_extractor.eval()

print("✓ ResNet-50 loaded")
print("✓ Classification layer removed")
print("Embedding size: 2048")


# ------------------------------------------------
# 4. Image preprocessing
# ------------------------------------------------

preprocess = transforms.Compose([
    transforms.Resize((224, 224)),

    transforms.Grayscale(num_output_channels=3),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ------------------------------------------------
# 5. Verify image paths
# ------------------------------------------------

missing_images = []

for path in df["image_path"]:

    if not os.path.exists(path):
        missing_images.append(path)

print("\nImage verification:")

print("Total images:", len(df))
print("Missing images:", len(missing_images))

if missing_images:

    print("\nFirst missing files:")
    print(missing_images[:10])

    raise FileNotFoundError(
        "Some JSRT images cannot be found."
    )

print("✓ All JSRT images found")


# ------------------------------------------------
# 6. Extract CNN embeddings
# ------------------------------------------------

print("\n" + "=" * 70)
print("EXTRACTING CNN EMBEDDINGS")
print("=" * 70)

embeddings = []

with torch.no_grad():

    for i, path in enumerate(df["image_path"]):

        try:

            image = Image.open(path).convert("L")

            image_tensor = preprocess(
                image
            ).unsqueeze(0).to(device)

            embedding = feature_extractor(
                image_tensor
            )

            embedding = embedding.squeeze()

            embedding = embedding.cpu().numpy()

            embeddings.append(embedding)

        except Exception as e:

            print(
                f"\nError processing image {i}: "
                f"{path}"
            )

            print(e)

            raise

        if (i + 1) % 10 == 0:

            print(
                f"Processed "
                f"{i + 1}/{len(df)}"
            )


# ------------------------------------------------
# 7. Convert to matrix
# ------------------------------------------------

X_cnn = np.vstack(embeddings)

print("\n" + "=" * 70)
print("CNN EMBEDDING MATRIX")
print("=" * 70)

print("\nShape:")
print(X_cnn.shape)

print("\nExpected:")
print("(154, 2048)")

assert X_cnn.shape == (len(df), 2048)

print("\n✓ Correct embedding dimensions")


# ------------------------------------------------
# 8. Save embeddings
# ------------------------------------------------

embedding_columns = [
    f"cnn_{i:04d}"
    for i in range(X_cnn.shape[1])
]

embedding_df = pd.DataFrame(
    X_cnn,
    columns=embedding_columns
)

embedding_df.insert(
    0,
    "study_id",
    df["study_id"].values
)

embedding_df["lung_cancer"] = (
    df["lung_cancer"].values
)

embedding_df["fold"] = (
    df["fold"].values
)


output_path = (
    "/kaggle/working/"
    "conan_jsrt_resnet50_embeddings.csv"
)

embedding_df.to_csv(
    output_path,
    index=False
)


# ------------------------------------------------
# 9. Final output
# ------------------------------------------------

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(output_path)

print("\nShape:")
print(embedding_df.shape)

print("\nColumns:")
print(
    f"study_id + "
    f"{len(embedding_columns)} CNN features + "
    f"lung_cancer + fold"
)

print("\n✓ CNN feature extraction complete")
```

```python
# ================================================================
# CONAN — CNN EMBEDDING 5-FOLD OOF CANCER RISK MODEL
# ================================================================

import os
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    roc_auc_score,
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("=" * 70)
print("CONAN — CNN EMBEDDING 5-FOLD OOF CANCER RISK MODEL")
print("=" * 70)

# ------------------------------------------------
# 1. LOAD CNN EMBEDDINGS
# ------------------------------------------------

path = "/kaggle/working/conan_jsrt_resnet50_embeddings.csv"

if not os.path.exists(path):
    raise FileNotFoundError(
        f"Could not find:\n{path}"
    )

df = pd.read_csv(path)

print("\nLoaded:")
print(path)

print("\nDataset shape:")
print(df.shape)

# ------------------------------------------------
# 2. IDENTIFY CNN FEATURES
# ------------------------------------------------

excluded = [
    "study_id",
    "lung_cancer",
    "fold"
]

feature_cols = [
    c for c in df.columns
    if c not in excluded
]

print("\nCNN feature count:")
print(len(feature_cols))

if len(feature_cols) != 2048:
    raise ValueError(
        f"Expected 2048 CNN features, found {len(feature_cols)}"
    )

X = df[feature_cols].astype(np.float32).values
y = df["lung_cancer"].astype(int).values
folds = df["fold"].astype(int).values

print("\nFeature matrix:")
print(X.shape)

print("\nTarget distribution:")
print(pd.Series(y).value_counts().sort_index())

print("\nFold distribution:")
print(
    pd.DataFrame({
        "fold": folds,
        "lung_cancer": y
    })
    .groupby(["fold", "lung_cancer"])
    .size()
    .unstack(fill_value=0)
)

# ------------------------------------------------
# 3. OOF PREDICTIONS
# ------------------------------------------------

oof_pred = np.zeros(len(df), dtype=float)
oof_class = np.zeros(len(df), dtype=int)

fold_results = []

print("\n" + "=" * 70)
print("TRAINING 5 FOLDS")
print("=" * 70)

for fold in sorted(np.unique(folds)):

    train_idx = folds != fold
    test_idx = folds == fold

    X_train = X[train_idx]
    X_test = X[test_idx]

    y_train = y[train_idx]
    y_test = y[test_idx]

    print(f"\nFold {fold}")
    print("-" * 50)
    print(f"Training images : {len(X_train)}")
    print(f"Validation images: {len(X_test)}")
    print(
        f"Train cancer    : {y_train.sum()} / {len(y_train)}"
    )
    print(
        f"Valid cancer    : {y_test.sum()} / {len(y_test)}"
    )

    # ------------------------------------------------
    # IMPORTANT:
    # Scaling is fitted ONLY on training data.
    # ------------------------------------------------

    model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                C=0.03,
                max_iter=3000,
                class_weight="balanced",
                solver="liblinear",
                random_state=42
            )
        )
    ])

    model.fit(X_train, y_train)

    prob = model.predict_proba(X_test)[:, 1]
    pred = (prob >= 0.5).astype(int)

    oof_pred[test_idx] = prob
    oof_class[test_idx] = pred

    fold_auc = roc_auc_score(y_test, prob)

    fold_results.append({
        "fold": fold,
        "n": len(y_test),
        "benign": int((y_test == 0).sum()),
        "malignant": int((y_test == 1).sum()),
        "roc_auc": fold_auc
    })

    print(f"Fold ROC-AUC: {fold_auc:.4f}")

# ------------------------------------------------
# 4. OVERALL OOF PERFORMANCE
# ------------------------------------------------

print("\n" + "=" * 70)
print("OVERALL 5-FOLD OOF PERFORMANCE")
print("=" * 70)

auc = roc_auc_score(y, oof_pred)
accuracy = accuracy_score(y, oof_class)
balanced_acc = balanced_accuracy_score(y, oof_class)
precision = precision_score(y, oof_class, zero_division=0)
recall = recall_score(y, oof_class, zero_division=0)
f1 = f1_score(y, oof_class, zero_division=0)

print(f"\nROC-AUC            : {auc:.4f}")
print(f"Accuracy           : {accuracy:.4f}")
print(f"Balanced Accuracy  : {balanced_acc:.4f}")
print(f"Precision          : {precision:.4f}")
print(f"Recall             : {recall:.4f}")
print(f"F1 Score           : {f1:.4f}")

# ------------------------------------------------
# 5. CONFUSION MATRIX
# ------------------------------------------------

cm = confusion_matrix(y, oof_class)

print("\nConfusion Matrix:")
print(cm)

# ------------------------------------------------
# 6. FOLD SUMMARY
# ------------------------------------------------

fold_df = pd.DataFrame(fold_results)

print("\nFold Results:")
print(fold_df.to_string(index=False))

print("\nFold ROC-AUC mean:")
print(f"{fold_df['roc_auc'].mean():.4f}")

print("\nFold ROC-AUC standard deviation:")
print(f"{fold_df['roc_auc'].std():.4f}")

# ------------------------------------------------
# 7. SAVE OOF PREDICTIONS
# ------------------------------------------------

oof_df = df[
    ["study_id", "lung_cancer", "fold"]
].copy()

oof_df["cnn_risk_probability"] = oof_pred
oof_df["cnn_prediction"] = oof_class

output_path = "/kaggle/working/conan_jsrt_cnn_oof_predictions.csv"

oof_df.to_csv(output_path, index=False)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(output_path)
print("Shape:", oof_df.shape)

print("\n✓ Every image has an out-of-fold CNN prediction")
print("✓ No validation image was used to train its fold model")
print("✓ CNN risk probability generated")
```

```python
# ================================================================
# CONAN — JSRT RESNET-50 FINE-TUNING
# 5-FOLD OUT-OF-FOLD LUNG CANCER RISK MODEL
# ================================================================

import os
import copy
import random
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

from sklearn.metrics import (
    roc_auc_score,
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# ================================================================
# CONFIGURATION
# ================================================================

SEED = 42

BATCH_SIZE = 8
NUM_EPOCHS = 15
LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-4

PATIENCE = 4

IMAGE_SIZE = 224

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("=" * 70)
print("CONAN — JSRT RESNET-50 FINE-TUNING")
print("=" * 70)

print("\nDevice:")
print(DEVICE)

# ------------------------------------------------
# Reproducibility
# ------------------------------------------------

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# ================================================================
# 1. LOAD JSRT DATA
# ================================================================

DATA_PATH = "/kaggle/working/conan_jsrt_5fold.csv"

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(
        f"Dataset not found:\n{DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

print("\nLoaded:")
print(DATA_PATH)

print("\nDataset shape:")
print(df.shape)

required_columns = [
    "study_id",
    "lung_cancer",
    "image_path",
    "fold"
]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(
            f"Required column missing: {col}"
        )

# ================================================================
# 2. VALIDATE DATA
# ================================================================

print("\n" + "=" * 70)
print("DATA VALIDATION")
print("=" * 70)

print("\nTarget distribution:")
print(df["lung_cancer"].value_counts().sort_index())

print("\nFold distribution:")
print(
    df.groupby(["fold", "lung_cancer"])
      .size()
      .unstack(fill_value=0)
)

# Check duplicate studies
duplicates = df["study_id"].duplicated().sum()

print("\nDuplicate study IDs:")
print(duplicates)

if duplicates != 0:
    raise ValueError("Duplicate study IDs detected.")

# Check image paths
missing_images = [
    p for p in df["image_path"]
    if not os.path.exists(p)
]

print("\nMissing images:")
print(len(missing_images))

if len(missing_images) > 0:
    print(missing_images[:10])
    raise FileNotFoundError(
        "Some JSRT images could not be found."
    )

print("\n✓ Dataset validated")
print("✓ One row per image")
print("✓ All images exist")
print("✓ No duplicate study IDs")

# ================================================================
# 3. DATASET CLASS
# ================================================================

class JSRTDataset(Dataset):

    def __init__(self, dataframe, transform=None):

        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        image_path = row["image_path"]
        label = int(row["lung_cancer"])

        # ------------------------------------------------
        # JSRT X-ray
        # ------------------------------------------------

        image = Image.open(image_path).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        return image, torch.tensor(
            label,
            dtype=torch.float32
        )

# ================================================================
# 4. IMAGE TRANSFORMS
# ================================================================

# Training augmentation
train_transform = transforms.Compose([

    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=5
    ),

    transforms.RandomAffine(
        degrees=0,
        translate=(0.03, 0.03),
        scale=(0.97, 1.03)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# Validation transform
val_transform = transforms.Compose([

    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("\n✓ Training augmentation configured")
print("✓ Validation transformation configured")

# ================================================================
# 5. RESNET-50 MODEL
# ================================================================

def create_model():

    print("\nLoading pretrained ResNet-50...")

    model = models.resnet50(
        weights=models.ResNet50_Weights.DEFAULT
    )

    # ------------------------------------------------
    # Freeze entire network initially
    # ------------------------------------------------

    for param in model.parameters():
        param.requires_grad = False

    # ------------------------------------------------
    # Fine-tune final convolutional block
    # ------------------------------------------------

    for param in model.layer4.parameters():
        param.requires_grad = True

    # ------------------------------------------------
    # Replace classifier
    # ------------------------------------------------

    num_features = model.fc.in_features

    model.fc = nn.Sequential(

        nn.Dropout(
            p=0.40
        ),

        nn.Linear(
            num_features,
            1
        )
    )

    # New classifier must train
    for param in model.fc.parameters():
        param.requires_grad = True

    return model.to(DEVICE)

# ================================================================
# 6. TRAINING FUNCTION
# ================================================================

def train_one_fold(
    train_df,
    val_df,
    fold_number
):

    print("\n" + "=" * 70)
    print(f"TRAINING FOLD {fold_number}")
    print("=" * 70)

    print(
        f"Training images   : {len(train_df)}"
    )

    print(
        f"Validation images : {len(val_df)}"
    )

    print(
        f"Training cancer   : {train_df['lung_cancer'].sum()} / {len(train_df)}"
    )

    print(
        f"Validation cancer : {val_df['lung_cancer'].sum()} / {len(val_df)}"
    )

    # ------------------------------------------------
    # Datasets
    # ------------------------------------------------

    train_dataset = JSRTDataset(
        train_df,
        transform=train_transform
    )

    val_dataset = JSRTDataset(
        val_df,
        transform=val_transform
    )

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=2,
        pin_memory=torch.cuda.is_available()
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=2,
        pin_memory=torch.cuda.is_available()
    )

    # ------------------------------------------------
    # Model
    # ------------------------------------------------

    model = create_model()

    # ------------------------------------------------
    # Class weighting
    # ------------------------------------------------

    train_labels = train_df["lung_cancer"].values

    n_negative = np.sum(train_labels == 0)
    n_positive = np.sum(train_labels == 1)

    pos_weight = (
        n_negative / n_positive
    )

    pos_weight_tensor = torch.tensor(
        [pos_weight],
        dtype=torch.float32,
        device=DEVICE
    )

    criterion = nn.BCEWithLogitsLoss(
        pos_weight=pos_weight_tensor
    )

    # ------------------------------------------------
    # Optimizer
    # ------------------------------------------------

    trainable_params = [
        p for p in model.parameters()
        if p.requires_grad
    ]

    optimizer = torch.optim.AdamW(
        trainable_params,
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.5,
        patience=2
    )

    # ------------------------------------------------
    # Early stopping
    # ------------------------------------------------

    best_auc = -np.inf
    best_state = None
    patience_counter = 0

    # ====================================================
    # EPOCH LOOP
    # ====================================================

    for epoch in range(NUM_EPOCHS):

        # ------------------------------------------------
        # TRAIN
        # ------------------------------------------------

        model.train()

        train_loss = 0.0

        for images, labels in train_loader:

            images = images.to(
                DEVICE,
                non_blocking=True
            )

            labels = labels.to(
                DEVICE,
                non_blocking=True
            ).unsqueeze(1)

            optimizer.zero_grad()

            logits = model(images)

            loss = criterion(
                logits,
                labels
            )

            loss.backward()

            optimizer.step()

            train_loss += (
                loss.item() * images.size(0)
            )

        train_loss /= len(train_dataset)

        # ------------------------------------------------
        # VALIDATION
        # ------------------------------------------------

        model.eval()

        val_probs = []
        val_labels = []

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(
                    DEVICE,
                    non_blocking=True
                )

                logits = model(images)

                probs = torch.sigmoid(
                    logits
                ).flatten()

                val_probs.extend(
                    probs.cpu().numpy()
                )

                val_labels.extend(
                    labels.numpy()
                )

        val_probs = np.asarray(
            val_probs
        )

        val_labels = np.asarray(
            val_labels
        ).astype(int)

        val_auc = roc_auc_score(
            val_labels,
            val_probs
        )

        scheduler.step(val_auc)

        print(
            f"Epoch {epoch + 1:02d}/{NUM_EPOCHS} | "
            f"Loss: {train_loss:.4f} | "
            f"Val AUC: {val_auc:.4f}"
        )

        # ------------------------------------------------
        # Save best model
        # ------------------------------------------------

        if val_auc > best_auc:

            best_auc = val_auc

            best_state = copy.deepcopy(
                model.state_dict()
            )

            patience_counter = 0

        else:

            patience_counter += 1

        if patience_counter >= PATIENCE:

            print(
                f"Early stopping at epoch "
                f"{epoch + 1}"
            )

            break

    # ====================================================
    # RESTORE BEST MODEL
    # ====================================================

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    print(
        f"\nBest validation ROC-AUC: "
        f"{best_auc:.4f}"
    )

    # ====================================================
    # FINAL VALIDATION PREDICTIONS
    # ====================================================

    model.eval()

    final_probs = []

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(
                DEVICE,
                non_blocking=True
            )

            logits = model(images)

            probs = torch.sigmoid(
                logits
            ).flatten()

            final_probs.extend(
                probs.cpu().numpy()
            )

    final_probs = np.asarray(
        final_probs
    )

    return final_probs, best_auc

# ================================================================
# 7. 5-FOLD OUT-OF-FOLD TRAINING
# ================================================================

print("\n" + "=" * 70)
print("5-FOLD OUT-OF-FOLD FINE-TUNING")
print("=" * 70)

oof_predictions = np.zeros(
    len(df),
    dtype=float
)

fold_results = []

unique_folds = sorted(
    df["fold"].unique()
)

for fold in unique_folds:

    train_df = df[
        df["fold"] != fold
    ].copy()

    val_df = df[
        df["fold"] == fold
    ].copy()

    val_probs, best_auc = train_one_fold(
        train_df,
        val_df,
        fold
    )

    # Put predictions back into original rows
    val_indices = val_df.index.values

    oof_predictions[
        val_indices
    ] = val_probs

    fold_results.append({

        "fold": int(fold),

        "n": len(val_df),

        "benign": int(
            (val_df["lung_cancer"] == 0).sum()
        ),

        "malignant": int(
            (val_df["lung_cancer"] == 1).sum()
        ),

        "roc_auc": best_auc
    })

# ================================================================
# 8. OVERALL OOF PERFORMANCE
# ================================================================

y_true = df[
    "lung_cancer"
].astype(int).values

oof_classes = (
    oof_predictions >= 0.5
).astype(int)

overall_auc = roc_auc_score(
    y_true,
    oof_predictions
)

accuracy = accuracy_score(
    y_true,
    oof_classes
)

balanced_accuracy = balanced_accuracy_score(
    y_true,
    oof_classes
)

precision = precision_score(
    y_true,
    oof_classes,
    zero_division=0
)

recall = recall_score(
    y_true,
    oof_classes,
    zero_division=0
)

f1 = f1_score(
    y_true,
    oof_classes,
    zero_division=0
)

cm = confusion_matrix(
    y_true,
    oof_classes
)

# ================================================================
# 9. RESULTS
# ================================================================

fold_df = pd.DataFrame(
    fold_results
)

print("\n" + "=" * 70)
print("FINAL 5-FOLD OOF RESULTS")
print("=" * 70)

print(
    f"\nOverall OOF ROC-AUC   : "
    f"{overall_auc:.4f}"
)

print(
    f"Accuracy              : "
    f"{accuracy:.4f}"
)

print(
    f"Balanced Accuracy     : "
    f"{balanced_accuracy:.4f}"
)

print(
    f"Precision             : "
    f"{precision:.4f}"
)

print(
    f"Recall                : "
    f"{recall:.4f}"
)

print(
    f"F1 Score              : "
    f"{f1:.4f}"
)

print("\nConfusion Matrix:")
print(cm)

print("\nFold Results:")
print(
    fold_df.to_string(
        index=False
    )
)

print(
    "\nMean Fold ROC-AUC: "
    f"{fold_df['roc_auc'].mean():.4f}"
)

print(
    "Fold ROC-AUC SD: "
    f"{fold_df['roc_auc'].std():.4f}"
)

# ================================================================
# 10. SAVE OOF PREDICTIONS
# ================================================================

output = df[
    [
        "study_id",
        "lung_cancer",
        "fold"
    ]
].copy()

output[
    "cnn_finetuned_risk_probability"
] = oof_predictions

output[
    "cnn_finetuned_prediction"
] = oof_classes

output_path = (
    "/kaggle/working/"
    "conan_jsrt_resnet50_finetuned_oof.csv"
)

output.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(output_path)
print("Shape:", output.shape)

print("\n✓ 5-fold OOF predictions generated")
print("✓ Validation images were not used for training")
print("✓ ResNet-50 layer4 was fine-tuned")
print("✓ Cancer probabilities generated")
print("✓ Results are ready for comparison")
```
